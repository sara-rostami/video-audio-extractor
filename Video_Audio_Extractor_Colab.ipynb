Downloads a video from a URL (via `yt-dlp` or a direct HTTP download) and extracts its audio track with `ffmpeg`.

```python
# 2) Enter the video URL and choose a method
video_url = input("Video URL: ").strip()
method = input("Method [auto/ytdlp/direct] (press Enter for auto): ").strip() or "auto"
audio_format = input("Audio format [mp3/wav/aac/flac] (press Enter for mp3): ").strip() or "mp3"
```

```python
# 3) Run the extractor
!python video_audio_extractor.py "{video_url}" --method {method} --audio-format {audio_format}
```

```python
# 4) Download the resulting audio file to your computer
import glob
from google.colab import files

audio_files = glob.glob("downloads/audio.*")
if audio_files:
    print(f"Downloading: {audio_files[0]}")
    files.download(audio_files[0])
else:
    print("No audio file found — check the logs above for errors.")
```
